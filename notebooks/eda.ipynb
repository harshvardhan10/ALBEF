```python
cam = np.array(pred_score_dist['cam_score'], dtype=np.float32)
cls = np.array(pred_score_dist['cls_score'], dtype=np.float32)
fused = np.array(pred_score_dist['fused_score'], dtype=np.float32)
```

```python
def _pct(x: np.ndarray, p: float) -> float:
    return float(np.percentile(x, p)) if x.size else float("nan")

def summarize_array(name: str, arr: np.ndarray):
        arr = np.asarray(arr, dtype=np.float32)
        print(
            f"{name:10s} n={arr.size:7d}  "
            f"min={arr.min():.6f}  "
            f"med={_pct(arr,50):.6f}  "
            f"p90={_pct(arr,90):.6f}  "
            f"p99={_pct(arr,99):.6f}  "
            f"max={arr.max():.6f}"
        )
```

```python
print("\n[Score Summary] Overall")
summarize_array("cam_score", cam)
summarize_array("cls_score", cls)
summarize_array("fused", fused)
```

```python
  # Compute counts per label
from collections import Counter
cnt = Counter([r["label"] for _, r in pred_score_dist.iterrows()])
labels_sorted = [l for l, _ in cnt.most_common(10)]

print(f"\n[Score Summary] Per-label (top {10} by #preds)")
for lb in labels_sorted:
    preds_lb = [r for _, r in pred_score_dist.iterrows() if r["label"] == lb]
    cam_lb = np.array([r["cam_score"] for r in preds_lb], dtype=np.float32)
    cls_lb = np.array([r["cls_score"] for r in preds_lb], dtype=np.float32)
    fused_lb = np.array([r["fused_score"] for r in preds_lb], dtype=np.float32)
    print(f"\nLabel: {lb}  (n={len(preds_lb)})")
    summarize_array("cam_score", cam_lb)
    summarize_array("cls_score", cls_lb)
    summarize_array("fused", fused_lb)
```
